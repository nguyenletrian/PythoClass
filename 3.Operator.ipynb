### ERROE HANDLING

```python
a = 10
b = 0
try:
    result = a / b
    print("Result:",result)
    print("Ket thuc")
except:
    print("Error: Chia cho 0")
print("Ket thuc")
```

```python
num = 10
print(num)
print(bin(num)) #0b 1010
```

```python
a = 5
b = 3
#Toan tu bitwise
print(a & b) #And

#Toan tu logical
print((a > 3) and (b > 5))
print(f"~{a} =",~a)
print(f"{a} & {b} =", a & b )
```

```python
a = 10
b = 5
# Toan tu so hoc (arithmetic): +,-,*,/,%,//,**
print(f"{a} + {b}",a + b)
print(f"{a} - {b}",a - b)
print(f"{a} * {b}",a * b)
print(f"{a} / {b}",a / b)
print(f"{a} // {b}",a // b)
print(f"{a} % {b}",a % b)
print(f"{a} ** {b}",a ** b)
```
