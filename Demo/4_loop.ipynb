# While

```python
# in 10 lan chu Hello dung while
count = 1
while count <= 5:
    print(f"{count}. Hello")
    count += 1

print("End")
```

# Nested loop

```python
# in matrix 3 dong` 2 cot
for i in range(1,4):
    for j in range(1,3):
        print(i,j)
```

```python
# in hinh vuong 4*4
'''
* * * * 
* * * * 
* * * * 
* * * * 
'''
#c1
for i in range(4):
    for j in range(4):
        print("*", end=" ")
    print()

print()

#c2
count = 4
for i in range(count):
    print("@ " * (count))
```

# break

```python
# break la ngừng vòng lặp
for i in range(1, 6):
    if i == 3:
        break
    print(i)

print("End")
```

```python
# continue la ko chay cac cau lenh trong lần lặp đó, 
# vẫn chạy tiếp các lần lặp sau đến hết vòng lặp
for i in range(1, 6):
    if i == 3:
        continue
    print(i)

print("End")
```

```python
#pass: ko lam gi, chi de giu cho
for i in range(6):
    pass

print("End")
```

```python
#else: chi chay khi vong for ko bi break
numbers = [1, 2, 3, 4, 5]
findnumber = 6
for i in numbers:
    if i == findnumber:
        print(f"{findnumber} co ton tai trong danh sach")
        break
else:
    print(f"{findnumber} khong ton tai trong danh sach")
```
