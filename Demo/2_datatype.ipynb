# Numeric Type

### Integer Type

```python
# khai bao bien kieu integer(so nguyen duong)
age = 10
print(age)
print(type(age))
```

```python
# khai bao bien kieu integer(so nguyen am)
num = -1
print(num)
print(type(num))
```

```python
# khai bao bien kieu integer (big number) (underscore)
big_num = 1_000_000_000
print(big_num)
print(type(big_num))
```

### Float type


```python
# khai bao bien kieu float(so thuc)
mark = 80.5
print("Diem cua ban la:", mark)
print("Diem cua ban la:" + str(mark))
print(f"Diem cua ban la: {mark}")
```

### Boolean Type

```python
# khai bao bien kieu bool
check = True
print("Check:", check)
print("Type: ", type(check))
```

### String type


```python
# khai bao bien kieu du lieu string
# double quote
name = "FPT Jetking"
age = 16
print("Ten: ", name, "-Tuoi: ", age)
print(f"Ten: {name}    -    Tuoi: {age}")
```

```python
# single quote
student_name = 'Thuy Vy'
print(student_name)
print(type(student_name))
```

```python
# multi quote
sentence = '''
    Today is Wednesday
    We learn session 2
'''
print(sentence)
print(type(sentence))
```

### String Index

```python
name = "Python Programming"
# lay 1 ky tu o vi tri dau tien 
print("name[0]:", name[0])
# lay 1 ky tu o vi tri cuoi cung
print("name[-1]:", name[-1])
# lay 1 ky tu o vi tri so 3
print("name[3]:", name[3])
```

### String Slicing

```python
name = "Python Programming"
print("name[0:6]: ", name[0:6]) #lay tu 0 toi 5
print("name[0:6:2]: ", name[0:6:2])
print("name[-6:-1]: ", name[-6:-1])
print("reverse: ", name[::-1]) 
```

### String operator

```python
# noi chuoi (concate)
name1 = "Python"
name2 = "Programming"
name = name1 + " " + name2
print(name)
```

```python
# lap chuoi
name1 = "Python"
name = name1 * 3
print(name)
```

```python
# membership
name1 = "Python"
check = "Py" in name1
print(check)
```

```python
check2 = "Java" not in name1
print(check2)
```

```python
# compare
name1 = "Python"
name2 = "Pylance"
check3 = name1 == name2
check4 = name1 < name2
print("check3:", check3)
print("check4:", check4)
```

### String Method

```python
# upper(): chuyen thanh chu hoa
name = "Python"
newName = name.upper()
print(name)
print(newName)
```

```python
# lower(): chuyen thanh chu thuong
lowerName = name.lower()
print(lowerName)
```

```python
# strip(): cat khoang trang o dau` va` cuoi' chuoi
str = "    Python   Programming    "
print(f"[{str}]")
print(f"[{str.strip()}]")
```

```python
# replace(): thay the chuoi con bang chuoi con moi'
string = "Hello Hello Python"
print(string.replace("llo", "lla"))
```

```python
# split(): cat chuoi thanh mang?, mac dinh la cat' theo dau khoang trang
print(string.split())
```

```python
fruits = "Apple, Banana, Water Melon"
print(fruits.split(","))
```
