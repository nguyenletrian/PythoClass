## Error Handling

```python
a = 10
b = 0
try:
    result = a/b
    print("result:", result)
except:
    print("Error: chia cho 0")

print("Ket thuc")
```

```python
a = 5
b = 3
# toan tu logical
print((a > 3) and (b > 5))

# toan tu bitwise
print(f"{a} & {b} = ",a & b)
print(f"~{a}=", ~a)
print()
```

```python
# toan tu so hoc(arithmetic): +, -, *, /, //, %, **
print(f"{a} + {b}=", a+b)
print(f"{a} - {b}=", a-b)
print(f"{a} * {b}=", a*b)
print(f"{a} / {b}=", a/b)
print(f"{a} // {b}=", a // b)
print(f"{a} % {b}=", a%b)
print(f"{a} ** {b}=", a**b)
```
