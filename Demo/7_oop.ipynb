# Class

```python
# khai bao 1 class
class Student:
    # khai bao 1 class attribute
    school_name = "FPT Jetking"

    # constructor: tu dong chay khi tao object
    # sname, smark la tham so cua phuong thuc init (special method)
    def __init__(self, sname, smark):
        self.name = sname # name chinh la instance attribute
        self.__mark = smark  # mark private instance attribute 

    # phuong thuc dac biet : __str__: dung de in doi tuong
    def __str__(self):
        return f"Student: name={self.name}, mark={self.__mark}"

    # instance method
    def display(self):
        print("Name: ", self.name, "--- Mark: ", self.__mark)

    # tao getter de doc du lieu cua bien __mark
    def get_mark(self):
        return self.__mark

    # tao setter de thay doi du lieu
    def set_mark(self, value):
        if 0 <= value <= 10:
            self.__mark = value
        else:
            print("Invalid mark")

# tao doi tuong(object)
student1 = Student("Tuan", 10)
student2 = Student("Khoa", 9)

# goi phuong thuc instance method
student1.display()
student2.display()

#  goi __str__
print(student1)
print(student2)

# in class attribute
print("School: ", Student.school_name)
print("School: ", student1.school_name)
print("School: ", student2.school_name)

# thay doi gia tri cua attribute
Student.school_name = "FPT"
student1.set_mark(9)

print("After change")
print(student1)
print("School: ", Student.school_name)
print("School: ", student1.school_name)
print("School: ", student2.school_name)
```
