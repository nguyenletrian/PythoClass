# Dictionary Create

```python
# tao 1 dictionary rong
empty_dict = {}
empty_dict2 = dict()
print("empty_dict: ", empty_dict)
print("empty_dict2: ", empty_dict2)
# tao 1 dictionary co du lieu
student = {"id": "SV01", "name": "Bao", "address": "Q12"}
print("student: ", student)

student2 = dict(id= "SV01", name= "Bao", address= "Q12")
print("student2: ", student2)
```

# Read

```python
# Truy cap theo key
print("Ma: ", student["id"])
print("Ten: ", student["name"])
print("Dia Chi: ", student["address"])
```

```python
# Duyet key
for key in student:
    print(key)
```

```python
# Duyet value
for value in student.values():
    print(value)
```

```python
# Duyet key va value
for key, value in student.items():
    print("Key: ", key, "--- Value: ", value)
```

# Add

```python
# them key moi
student["age"] = 18
print("student: ", student)
```

# Update

```python
# cap nhat lai tuoi
student["age"] = 19
print("student: ", student)
```

# Delete

```python
# xoa theo key
del student["age"]
print("student: ", student)
```

```python
student.pop("address")
print("student: ", student)
```

```python
# xoa het noi dung cua dict
student.clear()
print("student: ", student)
```

# Dictionary nested

```python
nested_dictionary = {
    "SV01": {'name': 'Bao', 'address': 'Q12'},
    "SV02":  {'name': 'Dat', 'address': 'Can Tho'},
    'SV03': {'name': 'Khoi', 'address': 'Go Vap'}
}

# duyet ds
print (nested_dictionary['SV03']['name'])
```

```python
for key, value in nested_dictionary.items():
    print("ID:", key, "--- Name:", value['name'], "--- Address:", value['address'])
```

## List chua moi phan tu la Dictionary

```python
student_list = [
    {'id': 'SV01', 'name': 'Bao', 'address': 'Q12'},
    {'id': 'SV02', 'name': 'Dat', 'address': 'Can Tho'},
    {'id': 'SV03', 'name': 'Khoi', 'address': 'Go Vap'},
    {'id': 'SV04', 'name': 'Khoa', 'address': 'Go Vap'}
]

for student in student_list:
    print("ID: ", student['id'], "--- Name:", student['name'], "--- Address:", student['address'])
```
