# Function Definition

```python
# ham ko co tham so
def greet():
    print("Hello")

# goi ham
greet()
```

```python
# ham co 1 tham so
def hello(message):
    print("Welcome:", message)

# goi ham
hello("Class")
hello(5)
hello()
```

```python
# ham co nhieu tham so
def add_numbers(num1, num2):
    sum = num1 + num2
    print("Sum: ", sum)

# goi ham
add_numbers(2, 3)
```

```python
# ham co return
def mul(num1, num2):
    result = num1 * num2
    return result

# goi ham
res = mul(2,3)
print("Res: ", res)
# neu nhu ko tao bien 
print("Result: ", mul(2,4))
```

```python
# keyword arguments
def add_numbers(num1, num2):
    sum = num1*2 + num2
    print("Sum: ", sum)

# goi ham
add_numbers(num1 = 2, num2 = 3)
add_numbers(num2 = 2, num1 = 3)
```

```python
# default arguments
def sub_numbers(num1, num2 = 3):
    result = num1 - num2
    print("Result: ", result)

# goi ham
sub_numbers(5)
sub_numbers(5, 2)
```

```python
# return multiple value
def calculate(numbers):
    total = sum(numbers)
    length = len(numbers)
    average = total/length if length > 0 else 0
    return total, length, average

# goi ham
data = [10, 20 , 30]
t, l, a = calculate(data)
print("Sum: ", t, " Length: ", l, " Average: ", a)
result = calculate(data)
print("Result: ", result)
```

# Scope: local, global

```python
# khai bao bien toan cuc (global)
count = 0

# dinh nghia 1 ham
def myFunc():
    # khai bao bien cuc bo (local)
    number = 1
    # khai bao 1 bien giong ten voi bien global o ngoai (no se tao 1 bien cuc bo moi)
    count = 1
    print("Inside function: ")
    print("number: ", number)
    print("count: ", count)

# goi ham
myFunc()
print("--------------")
print("Outside: count: ", count)
```

```python
# khai bao bien toan cuc (global)
count = 0

# dinh nghia 1 ham
def myFunc():
    # khai bao bien cuc bo (local)
    number = 1
    # muon doi gia tri cua bien global o ngoai thi phai khai bao
    global count
    count = 1
    print("Inside function: ")
    print("number: ", number)
    print("count: ", count)

# goi ham
myFunc()
print("--------------")
print("Outside: count: ", count)
```

# Lambda

```python
# lambda la ham an danh (khong can dat ten)
# dung de viet cac ham ngan gon
# co the co nhieu tham so, chi co 1 expression
add = lambda a,b:a+b
print("add: ", add(2,3))

square = lambda x: x**2
print("Square: ", square(3))
```

```python
# sorted(): ham tra ve list moi
numbers = [5,4,9,8,2,3]
sorted_list = sorted(numbers, key=lambda x:x)
print("sorted_list: ", sorted_list)

students = [("Tuan", 10), ("An", 9.5), ("Van", 0)]
student_list = sorted(students, key=lambda s:s[1])
print("student_list: ", student_list)
```

```python
#map(): tra ve 1 iterable sau khi thuc thi lambda
map_list = list(map(lambda x: x*2, numbers))
print("map_list:", map_list)
```

```python
# filter(): tra ve iterable, loc phan tu dua tren dieu kien cua lambda
filter_list= list(filter(lambda x: x%2==0, numbers))
print("filter_list:", filter_list)
```
