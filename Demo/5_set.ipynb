# Set Create

```python
empty_set = set()
student_set = {"Duc", "Khoa", "Dat", "Duc"}
print("empty_set: ", empty_set)
print("student_set: ", student_set)
```

```python
print(student_set[0]) # error vi ko truy cap theo vi tri duoc
```

```python
student_list = ["Tuan", "Khoa", "Tuan", "Duc"]
student_set2 = set(student_list)
print("student_list: ", student_list)
print("student_set2: ", student_set2)
```

```python
list = sorted(student_set) # sorted() la ham (function)
print("student_set: ", student_set)
print("list: ", list)
```

# Add and Remove Elements

```python
# add(item): them 1 phan tu vao set
student_set.add("Vy")
print("student_set: ", student_set)
```

```python
# update(iterator): them nhieu phan tu vao set
student_set.update(["Nhi", "Diem"])
print("student_set: ", student_set)
```

```python
# discard(item): xoa 1 phan tu neu phan tu ton tai trong set
# neu phan tu ko ton tai thi ko co loi xay ra
student_set.discard("Duc")
print("student_set: ", student_set)
student_set.discard("An")
print("student_set: ", student_set)
```

```python
#remove(item): xoa 1 phan tu neu phan tu ton tai trong set
# neu phan tu ko ton tai thi se bi bao loi
student_set.remove("Dat")
print("student_set: ", student_set)
student_set.remove("An")
print("student_set: ", student_set)
```

```python
# pop(): xoa 1 phan tu trong set
student_set.pop()
print("student_set: ", student_set)
```

```python
# clear(): xoa het cac phan tu trong set
student_set.clear()
print("student_set: ", student_set)
```

# Mathematic Operations

```python
set1 = {1, 2, 3, 4}
set2 = {3, 4, 5, 6}
# union(|)
print("set1 | set2 = ", set1 | set2)
# intersection(&)
print("set1 & set2 = ", set1 & set2)
# difference(-)
print("set1 - set2 = ", set1 - set2)
print("set2 | set1 = ", set2 - set1)
# symmetric difference(^)
print("set1 ^ set2 = ", set1 ^ set2)
# subset
print("{1, 2} <= set1 = ", {1,2} <= set1)
print("{1, 2}.issubset(set1) = ", {1,2}.issubset(set1))
# superset
print("{1, 2} >= set1 = ", {1,2} >= set1)
print("{1, 2, 3, 4, 5}.issuperset(set1) = ", {1, 2, 3, 4, 5}.issuperset(set1))
```
