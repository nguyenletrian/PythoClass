# if else

```python
# cho nhap 1 so , kiem tra xem so do la so chan hay so le
num = int(input("Nhap 1 so:"))
if num % 2 == 0:
    print(f"{num} la so chan")
else:
    print(f"{num} la so le")
 
print("ket thuc")
```

# if else if

```python
# cho nhap 1 diem, xuat ra phan loai dua tren diem
mark = float(input("Nhap diem cua sinh vien"))
if mark >= 90:
    print("Grade A")
elif mark >= 80:
    print("Grade B")
elif mark >= 70:
    print("Grade C")
elif mark >= 60:
    print("Grade D")
else:
    print("Grade E")

print("End")
```

```python
a = 5
b = 3
if a >= 5:
    print("ok")
if b >= 3:
    print("yes")
else:
    print("no")
```

## Nested if

```python
# VIP: revenue >= 1000 va year >= 5
revenue = 900
year = 5

if revenue >= 1000 and year >= 5 :
    print("VIP")
else:
    print("Not VIP")
```

```python
revenue = 1000
year = 4

if revenue >= 1000:
    if year >= 5:
        print("VIP")
    else:
        print("Not VIP because not enough year")
else:
    print("Not VIP because not enough revenue")
```
