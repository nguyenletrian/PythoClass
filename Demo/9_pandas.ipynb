```python
#inner join: chi lay cac ID xuat hien o ca 2 bang
df1 = pd.DataFrame({
    "ID": [1,2,3],
    "Product": ['Laptop', 'Phone', 'Mouse']
})

df2 = pd.DataFrame({
    "ID": [2,3,4],
    "Category": ['Cat1', 'Cat2', 'Cat3']
})

inner_join = pd.merge(df1, df2, on="ID", how="inner")
print(inner_join)
```

```python
# left join: chi lay phan ben trai
left_join = pd.merge(df1, df2, on="ID", how="left")
print(left_join)
```

```python
# right join: chi lay phan ben trai
right_join = pd.merge(df1, df2, on="ID", how="right")
print(right_join)
```

```python
# outer join: lay het ca 2 bang
outer_join = pd.merge(df1, df2, on="ID", how="outer")
print(outer_join)
```
