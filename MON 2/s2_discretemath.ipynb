# 2. Propositional Logic - Menh de

```python
# Truth Table (bang chan tri)
values = [True, False]
print("P \t Q \t AND \t OR \t NOT P \t P->Q")
for p in values:
    for q in values:
        result_and = p and q
        result_or = p or q
        result_not = not p
        implication = (not p) or q

        print(p, q, result_and, result_or, result_not, implication, sep=" \t")
```

# 3. Tautology - Menh de hang dung: P v not P

```python
values = [True, False]
for p in values:
    result = p or not p
    print ("P = ", p, "->", result)
```

# 8. Inclusive and Exclusive Principle -> Nguyen ly bu tru

```python
# 2 tap hop
total_students = 40
math = 25  # tap hop A
science = 20 # tap hop B
both = 10   # A giao(^) B
# tinh so hoc sinh chi thich Toan
only_math = math - both
# tinh so hoc sinh chi thich Khoa hoc
only_science = science - both
# tinh so hoc sinh thich it nhat mot mon(A U B)
# Cong thuc: n(A U B) = n(A) + n(B) - n(A^B)
either = math + science - both
# tinh so hoc sinh khong thich mon nao ca
none = total_students - either
print("So luong sv chi thich Toan: ", only_math)
print("So luong sv chi thich Khoa Hoc: ", only_science)
print("So luong sv thich it nhat 1 mon: ", either)
print("So luong sv ko thich mon nao: ", none)
```

```python
# 3 tap hop
total_students = 100
english = 40   # A
math = 50      # B
science = 30   # C
english_math = 10    # A^B
math_science = 8     # B^C
english_sciennce = 5 # A^C
like_three = 3  # A^B^C
# tinh so hoc sinh thich it nhat 1 mon
# A U B U C = A + B + C - (A^B) - (B^C) - (A^C) + (A^B^C)
any = english + math + science - english_math - math_science - english_sciennce + like_three
# tinh so hoc sinh ko thich mon nao
none = total_students - any
print("It nhat 1 mon: ", any)
print("Ko thich mon nao: ", none)
# tinh so hoc sinh chi thich dung' 2 mon english va math
only_english_math = english_math - like_three
print("thich dung 2 mon: ", only_english_math)
# tinh so hoc sinh chi thich dung' 1 mon math
only_math = math - english_math - math_science + like_three
print("dung 1 mon math: ", only_math)
```

# 9. Pascal Triangle (Binomial Coefficient)

```python
import math

rows = 5
for n in range(rows):  # n chay tu 0 den 4
    for k in range(n+1): # k chay tu 0 den n
        value = math.comb(n,k)
        print(f"{value} ", end="")
    print() # xuong dong` 
```

# 10. Permutation(Hoan vi)
Pₙᵏ= (n!)/(n-k)!

```python
# co quan tam toi thu tu
# sap xep
'''
Gia su co 3 chu so: 1, 2, 3
Tao password gom 2 chu so(phan biet)
Co may cach
'''
counts = math.perm(3,2)
print("Total ways:", counts)
```

```python
from itertools import permutations
# khai bao list cac so
digits = ["1", "2", "3"]
passwords = list(permutations(digits, 2))
print(passwords)
for p in passwords:
    print("".join(p))  # noi cac phan tu trong tuple thanh chuoi
print("Total ways: ", len(passwords))
```

# 11. Combination (To hop)
Cₙᵏ = n!/ (k!)(n-k)!

```python
# Co bao nhieu cach chon 5 cau thu chinh' tu 12 cau thu trong doi
total_players = 12
selected = 5
total_ways = math.comb(total_players, selected)
print("So cach: ", total_ways)
```

```python
from itertools import combinations

players = [f"P{i}" for i in range(1,13)]
print(players)
options = list(combinations(players, 5))
print("Tong so cach chon: ", len(options))
print("In 5 cach chon doi hinh dau tien:")
for op in options[0:5]:
    print(", ".join(op))
```

# 12. Graph: Adjacency List

```python
graph = {
    "A": ["B", "C", "D"],
    "B": ["A", "D"],
    "C": ["A"],
    "D": ["A", "B"]
}

for key, value in graph.items():
    print(key, "->", value)
```

```python
# tim ban cua anh A
# Vertices = User
# Edges = Friendship
person = "A"
print("Friends of ", person, "is " , graph[person])
```

# 13. Adjacency Matrix

```python
vertices = graph.keys()
print(vertices)
for v in vertices:
    row = []
    for o in vertices:
        if o in graph[v]:
            row.append(1)
        else:
            row.append(0)
    print(v, row)
```

# 20. Set Theory: Data Partitioning

```python
# tap du lieu gom 10 phan tu
Universal = set(range(1,11))
# phan hoach thanh tap Train va Test(ti le 70/30)
train = set([1, 2, 3, 4, 5, 6, 7])
test = set([8, 9, 10])
# kiem tra xem phan hoach dung chua bang cach su dung toan tu Giao(\cap ∩) va Hoi(\cup ∪)
intersect = train & test
union = train | test
print("intersect: ", intersect)
print("union: ", union)
print("Universal: ", Universal)
```

# 21. Set Theory - Classification

```python
numbers = [12, 7, 5, 22, 15, 9]
even = set()
odd = set()
# thuat toan phan loai
for num in numbers:
    if num % 2 == 0:
        even.add(num)
    else:
        odd.add(num)

print("Du lieu dau vao: ", numbers)
print("Tap so chan: ", even)
print("Tap so le:", odd)
```

# 22. Set Theory - Clustering

```python
ages = [5, 40, 8, 30, 35, 10]
# chon 2 tâm cụm la 6 va 32
center1 = 6
center2 = 32
cluster1 = set()
cluster2 = set()
for age in ages:
    dist1 = abs(age - center1)
    dist2 = abs(age - center2)
    if dist1 < dist2:
        cluster1.add(age)
    else:
        cluster2.add(age)

print("Cum 1 gan tam ", center1, ":", cluster1)
print("Cum 2 gan tam ", center2, ":", cluster2)
```

# 23. NLP & Text Analysis: chuan bi du lieu

```python
doc1 = "Mo hinh hoc may rat thong minh"
doc2 = "Hoc may la mot nhanh cua AI"
set1 = set(doc1.lower().split())
set2 = set(doc2.lower().split())
# hien thi tat ca tu vung cua 2 tap hop
all_word = set1 | set2
print("Tap tu vung cua tap 1: ", set1)
print("Tap tu vung cua tap 2: ", set2)
print("Tat ca tu vung: ", all_word)
```

# 24. Jaccard Similarity
jaccard = intersect/union

```python
intersect = set1 & set2
union = set1 | set2
jaccard = len(intersect)/len(union)
print(f"Do tuong dong cua 2 tap hop la: {jaccard*100:.2f} %")
```
