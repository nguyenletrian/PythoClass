```python
set1 = {"A","B","C","D"}
set2 = {"C","D","E","F"}
#union |  => tất cả thuộc 1 và 2
print("set 1 | set2 =",set1 | set2)

# intersection (&)
print("set 1 & set2 =", set1 & set2)

# diffserence(-) thuộc set 1 nhưng ko thuộc set 2
print("set1 - set2 =", set1 - set2)

# symmetric difference(^) # những phần tử của riêng mỗi set (- đi những phần tử giống)
print(set1 ^ set2)

#subset  <=  => # kiểm tra xem có phải là set có nằm trong set khác ko
print({1,2} <= set1)

#superset >=  có chứ 1 set khác ko
```
