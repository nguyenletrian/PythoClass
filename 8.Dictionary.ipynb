```python
ds = {
    'SV01':{'name':'Bao','address':'Q12'},
    'SV02':{'name':'Dat','address':'Can Tho'},
    'SV03':{'name':'Khoi','address':'Go Vap'},
}

# Print All
for key,val in ds.items():
    print(f"MSSV {key}: co ten la {val["name"]}, que o {val["address"]}")

# Print Khoi
for key,val in ds.items():
    if val["name"] == "Khoi":
        print(f"MSSV {key}: co ten la {val["name"]}, que o {val["address"]}")
        break
```
