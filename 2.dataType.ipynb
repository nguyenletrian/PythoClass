# NUMBERIC

## Interger Type

```python
# Khai bao kieu interger (so nguyen duong)
age = 10
print(age)
print(type(age))
```

```python
# Khai bao bien kieu integer (so nguyen am)
num = -1
print(num)
print(type(num))
```

```python
# Khai bao bien kieu interget (big number)
bigNum = 1_000_000_000
print(bigNum)
print(type(bigNum))
```

### FLOAT TYPE

```python
# Code khai bao bien kieu float (so thuc)
mark = 0.5
print("Diem cua ban la:",mark)
print("Diem cua ban la:" + str(mark))
print(f"Diem cua ban la: {mark}")
```

### BOOLEN TYPE

```python
# Khai bao bien kieu bool
check = True
print("Check:",check)
print("Type:",type(check))
```

### STRING TYPE

```python
# Khai bao bien kieu du lieu
name = "FPT Jetking"
age = 16
print("Ten: ",name,"-Tuoi:",age)
print(f"Ten: {name} - Tuoi: {age}")
```

```python
# Single quote
studenName = 'Tri An'
print(studenName)
print(type(studenName))
```

```python
# Multi quote
sentence = '''
    Today is Wednesday
    We learn session 2
'''
print(sentence)
print(type(sentence))
```

### STRING INDEX

```python
name = "Python Programming"
# Lay 1 ky tu o vi tri dau tien
print("name[0]:",name[0])

# Lay 1 ky tu o vi tri cuoi cung
print("name[-1]:",name[-1])

# Lay 1 ky tu o vi tri so 3
print('name[3]:',name[3])
```

### String Slicing

```python
name = "Python Programming"
print("name[0:6]:",name[0:6]) #lay tu 0 toi 5
print("name[0:6:2]: ",name[0:6:2])
print('name[-6:-1]: ',name[-6:-1])
print('name[-6:-1]: ',name[-6:])
print("reverse: ",name[::-1])
```

### STRING OPERATOR

```python
# Noi chuoi (concate)
name1 = "Python"
name2 = "Programming"
name = name1 +" "+ name2
print(name)
```

```python
# Lap chuoi
name1 = "Python"
name = name1*3
print(name)
```

```python
# Membership
name1 = "Python"
check = "Py" in name1
print(check)
```

```python
# Not membership
name1 = "Python"
check2 = "Java" not in name1
print(check2)
```

```python
# Compare
name1 = "Python"
name2 = "Pylance"
check3 = (name1 == name2)
check4 = name1 < name2
print("Check3:",check3)
print("check4:",check4)# t > l mã ASCII
```

### STRING METHODE

```python
# upper(): chuyen thanh chu hoa
name = "Python"
newName = name.upper()
print(name)
print(newName)
```

```python
# lower(): chuyen thanh chu thuong
name = "Python"
lowerName = name.lower()
print(lowerName)
```

```python
# strip(): cat khoang trang o dau va cuoi chuoi
string = "   Python     "
print(f"({string.strip()})")
```

```python
# replace(): Thay the chuoi con bang chuoi moi
string = "Hello Hello Python"
print(string.replace("Hello ",""))
```

```python
# Split(): cat chuoi thanh mang
print(string.split())
```

```python
fruits = "Apple,Banna,Water Melon "
print(fruits.split(","))
```
