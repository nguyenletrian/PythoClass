### WHILE LOOP

```python
count = 1
while count <= 5:
    print(count)
    count += 1
```

```python
# in 10 lan chu hello dung while
count = 1
while count <= 10:
    print(f"Hello {count}")
    count += 1
print("End")
```

```python
# Print range
for i in range(1,6): #từ 1 đến 5 , nếu chỉ truyền vào 1 tham số 6 thì nó sẽ chạy từ 0-5
    print(i)
```

```python
# Print chuỗi
for char in "Hello python":
    print(char)
```

```python
# In cac phan tu trong list
number = [1,2,3,4,5]
for item in number:
    print(item)
```

```python
list = ["Vy","Tuan","Khoa"]
for s in list:
    print(s)
print(s)
```

```python
count = 4
for i in range(0,count):
    print("*  " * (count))
```

```python
#Break
for i in range(1,10):
    if i ==5:
        break
    print(i)
```

```python
# PASS
for i in range(1,4):
    pass #Code to be added later
print("Loop completed")
```

```python
# CONTINUE
for i in range(1,6):
    if i == 3:
        continue
    print(i)
```

```python
# ELSE
for i in range(1,6):
    if i == 10:
        break
    print(i)
else:
    print("Loop completed successfully")
```

```python
# Else: Chỉ chạy khi vòng for ko bị break
number = [1,2,3,4,5,6]
findNumber = 3
for i in number:
    if i == findNumber:
        print(f"{findNumber} có ton tai trong danh sach")
        break
else:
    print(f"{findNumber} khong ton tai trong danh sach")
```

```python
for i in range(1,5):
    for j in range(1, 5):
        if j > i :
            continue
        print(j,end=" ")
    print()
```

```python
for i in range(7):
    listTemp = range(1,i)
    print((" ").join(map(str,listTemp)))
```
