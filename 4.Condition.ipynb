### If

```python
#Kiem tra cho nhap 1 so, kiem tra xem so do co phai so chan hay ko
num = input("Vui long nhap 1 so")
if num != "":
    num = int(num)   
    if (num % 2) == 0:
        print(f"{num} là số chẵn")
    else:
        print(f"{num} là số lẽ")
else:
    print("Khong hop le")
```

### IF ELSE IF

```python
# Cho nhap 1 diem, xuat ra phan loai dua tren diem
mark = float(input("Nhap diem cua sinh vien"))
if mark >= 90:
    print("Grade A")
elif mark >= 80:
    print("Grade B")
elif mark >= 70:
    print("Grade C")
elif mark >= 60:
    print("Grade D")
else:
    print("Grade E")
print("End")
```

```python
a = 5
b = 3
if a >= 5:
    print("ok")
if b >= 3:
    print("yes")
else:
    print("no")
```

### NESTED IF

```python
# VIP: revenue >= 1000 và year >=5
revenue = 1000
year = 5
if revenue >= 1000 and year >= 5:
    print("VIP")
else:
    print("Not VIP")
```

```python
# VIP: revenue >= 1000 và year >=5
revenue = 1000
year = 5
if revenue >= 1000:
    if year >= 5:
        print("VIP")
    else:
        print("Not VIP beacuse not enough year")
else:
    print("Not VIP because not enough revenue")
```
